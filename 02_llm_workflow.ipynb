```python
from langgraph.graph import START, END, StateGraph
from langchain_google_genai import ChatGoogleGenerativeAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash"
)
```

```python
class LLMState(TypedDict):

    question : str
    answer : str
```

```python
def llm_qa(state: LLMState) -> LLMState:

    # Extract Question
    question = state['question']

    #form a prompt
    prompt = f"Answer the following question in few words {question}"

    # ask that question
    answer = model.invoke(prompt).content

    #update the answer state
    state["answer"] = answer

    return state
```

```python
# graph initialize
graph = StateGraph(LLMState)

#Add node
graph.add_node('llm_qa',llm_qa)

# Add Edge
graph.add_edge(START, 'llm_qa')
graph.add_edge('llm_qa',END)

#compile
llm_workflow = graph.compile()
```

```python
initial_state = {'question' : 'How does the langgrapgh works?'}
final_state = llm_workflow.invoke(initial_state)

print(final_state['answer'])
```
